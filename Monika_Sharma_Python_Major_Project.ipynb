```python
# project for student registration

students = []  #list to register student data
```

```python
def accept():
    name = input("entter name:")
    roll = int(input("enter roll number:"))
    marks1 = int(input("enter marks1:"))
    marks2 = int(input("enter marks2:"))
    student = {'name':name, 'roll':roll, 'marks1':marks1, 'marks2':marks2}
    students.append(student)
    print("students registered successfully.")
```

```python
def display():
    if len(students) == 0:
        print("no student registered yet.")
    else:
        print("registered students:")
        for student in students:
            print("name:",student['name'])
            print("roll number:",student['roll'])
            print("marks1:",student['marks1'])
            print("marks2:",student['marks2'])
            print("***************************")
```

```python
def search():
    roll = int(input("enter roll number to search:"))
    for student in students:
        if student['roll'] == roll:
            print("student found:")
            print("name:",student['name'])
            print("roll number:",student['roll'])
            print("marks1:",student['marks1'])
            print("marks2:",student['marks2'])
            return
    print("student not found.")
```

```python
def delete():
    roll = int(input("enter roll number to delete:"))
    for student in students:
        if student['roll'] == roll:
            students.remove(student)
            print("student deleted successfully.")
            return
    print("student not found.")
```

```python
def update():
    roll = int(input("enter a roll number to update:"))
    for student in students:
        if student['roll'] == roll:
            name = input("enter new name:")
            marks1 = int(input("enter new marks1:"))
            marks2 = int(input("enter new marks2:"))
            student['name'] = name
            student['marks1'] = marks1
            student['marks2'] = marks2
            print("student updated successfully.")
            return
    print("student not found.")
```

```python
while True:
    print("1. Register student")
    print("2. Display students")
    print("3. search student")
    print("4. Delete student")
    print("5. Update student")
    print("6. Exit")
    
    choice = int(input("enter your choice(1-6):"))
    
    if choice == 1:
        accept()
    elif choice == 2:
        display()
    elif choice == 3:
        search()
    elif choice == 4:
        delete()
    elif choice == 5:
        update()
    elif choice == 6:
        print("Exiting program.")
        break
    else:
        print("Invalid choice. please try again.")
```
